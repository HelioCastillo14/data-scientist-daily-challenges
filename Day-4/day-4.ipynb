# Salud pública: intervalo de confianza para la cobertura de vacunación

Tipo: Matemática Aplicada · Nivel: 1 · Apoyo: Guiado · Tema: Inferencia estadística · Etapa: evaluación · Tiempo objetivo: 20-30 min (leer la teoría y las pistas NO cuenta en Tiempo Real)


## 📦 Datos

- Tamaño de la muestra: $n = 150$ hogares encuestados aleatoriamente.
- Hogares con esquema de vacunación completo: $x = 96$.
- Nivel de confianza deseado: 95% (usa $z = 1.96$; no lo calcules con tablas ni funciones).
- Meta institucional de cobertura: 70%.

1. Calcula la proporción muestral $\hat{p}$ dividiendo los vacunados encuestados entre el total de la muestra.

```python
n = 150 #Representa la muestra de la población
x = 96 #Representa el numero de Hogares que cumplen el esquema de vacunación completo -> caso de exito 

p_hat = x/n
print('La proporción muestral es: ', p_hat)
```

2. Verifica explícitamente la condición de aproximación normal con los valores de $n$ y $\hat{p}$ de este reto.

```python
if ((n*p_hat >=5) and (n*(1 - p_hat) >=5)):
    print('Se puede utilizar la aproximación normal')
else:
    print('No se puede utilizar la aproximación normal')
```

3. Calcula el error estándar con la fórmula de la Teoría aplicada, usando la raíz cuadrada (por ejemplo `math.sqrt` o `** 0.5`).

```python
import math

se = math.sqrt((p_hat*(1-p_hat))/n)

print('El error estándar es: ', round(se, 4))
```

El *error estandar* muestra una variabilidad tipica de **3 puntos porcentuales**. Eso quiere decir que nuestra muestra porcentual va a estar variando entre 0.68 y 0.60. 

4. Calcula el margen de error multiplicando el error estándar por $z=1.96$.

```python
z = 1.96 

margen = z * se
print('El margen de error es: ', round(margen, 4))
```

5. Calcula el límite inferior y el límite superior del intervalo de confianza, redondeados a 4 decimales.

```python
lo_ci = p_hat - margen 
hi_ci = p_hat + margen

print('El intervalo de confianza es: ', '[', round(lo_ci, 4), ' a ', round(hi_ci, 4), ']')
```

6. Compara el límite inferior del intervalo con la meta de cobertura del 70% y redacta la recomendación (extender o no extender la campaña).

```python
diferencia = hi_ci - 0.70
print('La diferencia entre el límite superior del intervalo de confianza y la proporción poblacional meta (70%) es: ', round(diferencia, 4))
```

7. Verifica tu desarrollo con los criterios del Entregable antes de marcar el reto como terminado.

```python
assert round(p_hat, 4) == 0.64
assert round(se, 4) == 0.0392
assert round(margen, 4) == 0.0768
assert round(lo_ci, 4) == 0.5632
assert round(hi_ci, 4) == 0.7168
print(round(p_hat, 4), round(se, 4), round(margen, 4), round(lo_ci, 4), round(hi_ci, 4))
```

# Recomendacion 

No hay evidencia estadística para extender la campaña, ya que la meta del 70% cae dentro de nuestro intervalo de confianza. Sin embargo, dado que estamos muy cercanos al límite superior, recomendamos monitorear continuamente la cobertura para asegurar que se mantiene en la meta, y considerar una expansión si futuras muestras muestran una tendencia descendente.
